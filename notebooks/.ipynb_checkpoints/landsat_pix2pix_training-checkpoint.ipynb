## 📌 Overview
This notebook trains and evaluates an end-to-end **Conditional Generative Adversarial Network (Pix2Pix)** that translates **Landsat Collection 2 Level-2 Thermal Infrared imagery (Band 10 Surface Temperature)** into photorealistic **Visible True-Color RGB imagery (Bands 4, 3, 2 Surface Reflectance)**.

### Pipeline Steps Covered:
1. **Environment Setup & Device Detection** (CUDA GPU / CPU)
2. **Radiometric Calibration & Dataset Preparation** (USGS Collection 2 Level-2 formulas + Scene-Level Splitting)
3. **Visual Data Exploration** (Interactive Matplotlib rendering of paired Thermal ↔ RGB patches)
4. **Model Architecture** (8-layer U-Net 256 Generator + 70x70 PatchGAN Discriminator)
5. **Pix2Pix Training Loop** (Dual Adam optimizers, Automatic Mixed Precision, Live loss logging)
6. **Full-Reference Scientific Evaluation** (PSNR, SSIM, MAE, Tenengrad Sharpness, Shannon Entropy, EPI)
7. **Visual Comparison Panels** (4-panel diagnostic side-by-side figures)
8. **Full-Scene Tiled Inference** (Seamless 2D Hann window blending)

## 1. Setup & Environment Verification

```python
import os
import sys
import time
import json
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn

# Ensure project root is on Python path
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

print("=" * 60)
print(f"PyTorch Version: {torch.__version__}")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Active Device:   {device}")
if torch.cuda.is_available():
    print(f"GPU Name:        {torch.cuda.get_device_name(0)}")
    print(f"VRAM Available:  {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
print("=" * 60)
```

## 2. Dataset Preparation & Patch Generation
We extract $256 \times 256$ paired patches. If you do not have raw Landsat GeoTIFFs downloaded yet, the cell below will automatically generate synthetic benchmark remote-sensing scenes so you can run and verify the complete training pipeline immediately.

```python
from src.data.patch_extractor import build_scene_level_dataset
from src.data.dataset import get_dataloaders

DATA_DIR = os.path.join(PROJECT_ROOT, "data")
os.makedirs(DATA_DIR, exist_ok=True)

# Ensure training/val/test patch directories exist
for split in ["train", "val", "test"]:
    os.makedirs(os.path.join(DATA_DIR, split, "ir"), exist_ok=True)
    os.makedirs(os.path.join(DATA_DIR, split, "rgb"), exist_ok=True)

# Check if dataset already has patches; if not, generate benchmark training pairs
train_ir_files = os.listdir(os.path.join(DATA_DIR, "train", "ir"))
if len(train_ir_files) < 10:
    print("Generating benchmark paired patches for training/val/test...")
    np.random.seed(42)
    splits_count = {"train": 60, "val": 15, "test": 15}
    for split, count in splits_count.items():
        for i in range(count):
            # Synthetic thermal surface temp normalized to [-1.0, 1.0]
            thermal_patch = np.random.uniform(-0.8, 0.8, (256, 256)).astype(np.float32)
            # Synthetic paired visible RGB in [-1.0, 1.0]
            rgb_patch = np.stack([
                thermal_patch * 0.7 + np.random.normal(0, 0.1, (256, 256)),
                thermal_patch * 0.9 + np.random.normal(0, 0.1, (256, 256)),
                thermal_patch * 0.5 + np.random.normal(0, 0.1, (256, 256))
            ], axis=-1).astype(np.float32)
            rgb_patch = np.clip(rgb_patch, -1.0, 1.0)

            np.save(os.path.join(DATA_DIR, split, "ir", f"scene_{split}_{i:04d}_ir.npy"), thermal_patch)
            np.save(os.path.join(DATA_DIR, split, "rgb", f"scene_{split}_{i:04d}_rgb.npy"), rgb_patch)

    print(f"✅ Generated {splits_count['train']} train, {splits_count['val']} val, and {splits_count['test']} test patches.")
else:
    print(f"✅ Dataset found with {len(train_ir_files)} training patches.")

# Create PyTorch DataLoaders
batch_size = 8 if torch.cuda.is_available() else 4
loaders = get_dataloaders(data_dir=DATA_DIR, batch_size=batch_size, num_workers=0)
print(f"Train batches: {len(loaders['train'])}, Val batches: {len(loaders['val'])}, Test batches: {len(loaders['test'])}")
```

## 3. Visual Exploration of Paired Dataset
Let's inspect a batch of paired **Thermal IR** input tensors and ground-truth **Visible RGB** target tensors.

```python
# Fetch one batch from DataLoader
sample_ir, sample_rgb = next(iter(loaders["train"]))

# Convert from [-1.0, 1.0] -> [0.0, 1.0] for Matplotlib display
def unnorm_ir(tensor):
    return ((tensor.squeeze().numpy() + 1.0) / 2.0).clip(0, 1)

def unnorm_rgb(tensor):
    return ((tensor.permute(1, 2, 0).numpy() + 1.0) / 2.0).clip(0, 1)

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for i in range(min(4, sample_ir.shape[0])):
    # Row 0: Thermal IR Input
    axes[0, i].imshow(unnorm_ir(sample_ir[i]), cmap="magma")
    axes[0, i].set_title(f"Thermal IR (Sample {i+1})", fontsize=11, fontweight="bold")
    axes[0, i].axis("off")
    
    # Row 1: Visible RGB Ground Truth
    axes[1, i].imshow(unnorm_rgb(sample_rgb[i]))
    axes[1, i].set_title(f"Ground Truth RGB (Sample {i+1})", fontsize=11, fontweight="bold")
    axes[1, i].axis("off")

plt.suptitle("Paired Landsat Dataset Inspection (Thermal Band 10 ↔ Visible Bands 4,3,2)", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

## 4. Model Architecture: U-Net Generator & PatchGAN Discriminator

```python
from src.models.generator import UNetGenerator
from src.models.discriminator import PatchGANDiscriminator
from src.models.losses import Pix2PixLoss

# 1. U-Net 256 Generator (1-channel IR In -> 3-channel RGB Out)
net_g = UNetGenerator(in_channels=1, out_channels=3, num_filters=64).to(device)

# 2. 70x70 PatchGAN Discriminator (4-channel [IR + RGB] In -> 30x30 Patch Out)
net_d = PatchGANDiscriminator(in_channels=4, num_filters=64).to(device)

# 3. Composite Loss Function (Adversarial BCE + 100 * L1 Reconstruction)
criterion = Pix2PixLoss(lambda_l1=100.0).to(device)

# Count model parameters
g_params = sum(p.numel() for p in net_g.parameters() if p.requires_grad)
d_params = sum(p.numel() for p in net_d.parameters() if p.requires_grad)
print(f"✅ U-Net Generator Parameters:     {g_params:,}")
print(f"✅ PatchGAN Discriminator Params: {d_params:,}")
print(f"✅ Total Trainable Parameters:    {g_params + d_params:,}")
```

## 5. Pix2Pix Training Loop
Run training with Automatic Mixed Precision (AMP) and dual Adam optimizers ($lr=0.0002, \beta_1=0.5, \beta_2=0.999$).

```python
from src.training.train import Pix2PixTrainer

CHECKPOINT_DIR = os.path.join(PROJECT_ROOT, "checkpoints")
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

# Choose number of epochs (e.g., 10 for quick Jupyter demo, 50 for full convergence)
NUM_EPOCHS = 10

trainer = Pix2PixTrainer(
    net_g=net_g,
    net_d=net_d,
    train_loader=loaders["train"],
    val_loader=loaders["val"],
    lr=0.0002,
    lambda_l1=100.0,
    checkpoint_dir=CHECKPOINT_DIR,
    device=device,
    use_amp=(device.type == "cuda")
)

print(f"🚀 Starting Pix2Pix Training on {device} for {NUM_EPOCHS} epochs...")
history = trainer.train(num_epochs=NUM_EPOCHS, save_freq=2)
```

## 6. Training Loss Dynamics & Convergence Curves

```python
epochs_range = range(1, len(history["loss_g"]) + 1)

plt.figure(figsize=(14, 5))

# Generator and Discriminator Loss
plt.subplot(1, 2, 1)
plt.plot(epochs_range, history["loss_g"], label="Generator Total Loss", color="#FF5A36", lw=2)
plt.plot(epochs_range, history["loss_d"], label="Discriminator Loss", color="#00D2FF", lw=2)
plt.title("Pix2Pix Adversarial Loss Progression", fontsize=12, fontweight="bold")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True, alpha=0.3)

# Validation L1 Loss Progression
plt.subplot(1, 2, 2)
plt.plot(epochs_range, history["val_l1"], label="Validation L1 Reconstruction Error", color="#10B981", lw=2)
plt.title("Validation Reconstruction Error (L1)", fontsize=12, fontweight="bold")
plt.xlabel("Epoch")
plt.ylabel("Mean Absolute Error")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 7. Scientific Evaluation on Held-Out Test Scenes
We evaluate the best model checkpoint on the test set, computing **PSNR**, **SSIM**, **MAE**, **Tenengrad Sharpness**, **Shannon Information Entropy**, and the **Edge Preservation Index (EPI)**.

```python
from src.evaluation.evaluate import evaluate_model_on_test_set

best_checkpoint = os.path.join(CHECKPOINT_DIR, "generator_best.pth")
eval_output_dir = os.path.join(PROJECT_ROOT, "outputs", "evaluation")

eval_results = evaluate_model_on_test_set(
    checkpoint_path=best_checkpoint,
    data_dir=DATA_DIR,
    output_dir=eval_output_dir,
    num_samples_to_save=4,
    device=str(device)
)

print("\n" + "=" * 60)
print("🎯 FULL-REFERENCE TEST EVALUATION SUMMARY")
print("=" * 60)
metrics = eval_results["metrics"]
print(f"• Peak Signal-to-Noise Ratio (PSNR): {metrics['psnr_mean_db']:.2f} dB (± {metrics['psnr_std_db']:.2f})")
print(f"• Structural Similarity Index (SSIM):  {metrics['ssim_mean']:.4f} (± {metrics['ssim_std']:.4f})")
print(f"• Mean Absolute Error (MAE):          {metrics['mae_mean_pixels']:.2f} pixels")
print(f"• Edge Preservation Index (EPI):      {metrics['epi_mean']:.4f}")
print(f"• Shannon Information Entropy:        {metrics['entropy_mean_bits']:.2f} bits")
print(f"• Tenengrad Edge Gradient:            {metrics['tenengrad_mean']:.2f}")
print("=" * 60)
```

## 8. Visual 4-Panel Diagnostic Figures
Displaying diagnostic comparisons:
$$\text{[Input Thermal IR \ \vert\ \ Predicted RGB \ \vert\ \ Ground Truth RGB \ \vert\ \ Heat Error Map]}$$

```python
import cv2

eval_images = [os.path.join(eval_output_dir, f) for f in os.listdir(eval_output_dir) if f.endswith(".png")]
if eval_images:
    for img_path in eval_images[:2]:
        panel_img = cv2.imread(img_path)
        panel_rgb = cv2.cvtColor(panel_img, cv2.COLOR_BGR2RGB)
        
        plt.figure(figsize=(16, 6))
        plt.imshow(panel_rgb)
        plt.title(f"Evaluation Panel: {os.path.basename(img_path)}", fontsize=12, fontweight="bold")
        plt.axis("off")
        plt.tight_layout()
        plt.show()
```

## 9. Full-Scene Tiled Sliding-Window Inference Demo
Seamlessly translate arbitrary-sized Landsat thermal images using 2D Hann window blending (zero boundary seams).

```python
from src.inference.predict import predict_thermal_array

# Create an arbitrary large thermal scene (e.g., 600 x 800)
large_thermal = np.random.uniform(-1.0, 1.0, (600, 800)).astype(np.float32)

# Run seamless sliding-window tiled inference
t0 = time.time()
stitched_rgb = predict_thermal_array(
    thermal_normalized=large_thermal,
    net_g=net_g,
    patch_size=256,
    overlap=64,
    device=str(device)
)
elapsed = time.time() - t0

print(f"✅ Full-Scene Tiled Inference Complete in {elapsed:.3f} seconds!")
print(f"Input Dimensions:  {large_thermal.shape} float32")
print(f"Output Dimensions: {stitched_rgb.shape} uint8 RGB")

# Plot result
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.imshow(large_thermal, cmap="magma")
plt.title(f"Large Thermal Input ({large_thermal.shape[1]}x{large_thermal.shape[0]})")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(stitched_rgb)
plt.title(f"Seamless Stitched RGB Output ({stitched_rgb.shape[1]}x{stitched_rgb.shape[0]})")
plt.axis("off")

plt.tight_layout()
plt.show()
```
